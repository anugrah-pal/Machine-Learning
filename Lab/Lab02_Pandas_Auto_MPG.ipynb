# 🧮 **3. Filtering / Boolean Indexing**

**Q3.1 Filter and display all cars with `mpg` greater than 30.**

# 🧹 **5. Handling Missing Values**

### _`.isnull().sum()` → count missing values per column | `.fillna(value)` → fill missing | `.dropna()` → drop rows/cols with missing values_

---

**Q5.1 Check how many missing values exist in each column of `df`.**

```python
df.isnull().sum()
```

**Q5.2 Fill any missing values in the `horsepower` column with the column's mean value.**

```python
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].mean())
print("Missing in horsepower after fill:", df["horsepower"].isnull().sum())
```

**Q5.3 Find the total number of missing values in the entire dataset (a single number) using `.isnull().sum().sum()`.**

```python
total_missing = df.isnull().sum().sum()
print("Total missing values in dataset:", total_missing)
```

# 🛠️ **6. Adding, Transforming & Dropping Columns**

### _Create a column by assignment: `df['new'] = ...` | `.apply(func)` → apply a function/lambda element-wise | `.drop(columns=[...])` → remove columns_

---

**Q6.1 Add a new column `weight_kg` by converting `weight` (in lbs) to kilograms (1 lb = 0.4536 kg).**

```python
df["weight_kg"] = df["weight"] * 0.4536
df[["weight", "weight_kg"]].head()
```

**Q6.2 Add a new column `efficiency` using `.apply()` with a lambda: label a car `'Good'` if `mpg >= 25`, `'Average'` if `15 <= mpg < 25`, else `'Poor'`.**

```python
df["efficiency"] = df["mpg"].apply(lambda x: "Good" if x >= 25 else ("Average" if x >= 15 else "Poor"))
print(df["efficiency"].value_counts())
```

**Q6.3 Drop the `name` column and store the result in a new DataFrame called `df_numeric`.**

```python
df_numeric = df.drop(columns=["name"])
df_numeric.head()
```

**Q6.4 Add a new column `hp_per_weight` by dividing `horsepower` by `weight` for every row.**

```python
df["hp_per_weight"] = df["horsepower"] / df["weight"]
df[["horsepower", "weight", "hp_per_weight"]].head()
```

# 📊 **7. GroupBy & Aggregation**

### _`.groupby('col')` → split by group | `.mean()`, `.max()`, `.count()` → aggregate | `.agg({...})` → different aggregation per column_

---

**Q7.1 Group by `origin` and find the average `mpg` for each origin.**

```python
df.groupby("origin")["mpg"].mean()
```

**Q7.2 Group by `cylinders` and find the average `mpg` and the maximum `horsepower` for each group using `.agg()`.**

```python
df.groupby("cylinders").agg({"mpg": "mean", "horsepower": "max"})
```

**Q7.3 Group by `model_year` and count how many cars were made each year.**

```python
df.groupby("model_year")["name"].count()
```

**Q7.4 Group by both `origin` and `cylinders` together, and find the average `mpg` for each combination.**

```python
df.groupby(["origin", "cylinders"])["mpg"].mean()
```

# 🔤 **9. String Operations**

# 🧾 **10. Duplicates & Renaming**

### _`.duplicated()` → boolean mask of duplicate rows | `.drop_duplicates()` → remove them | `.rename(columns={...})` → rename columns_

---

**Q10.1 Check whether `df` has any duplicate rows using `.duplicated()`, then remove them with `.drop_duplicates()`.**

```python
print("Duplicate rows count:", df.duplicated().sum())
df = df.drop_duplicates()
```

**Q10.2 Rename the `model_year` column to `year` and the `horsepower` column to `hp`.**

```python
df = df.rename(columns={"model_year": "year", "horsepower": "hp"})
print(df.columns)
```

**Q10.3 Rename every column name to uppercase in one line, using `.rename(columns=...)` with a dictionary comprehension.**

```python
df = df.rename(columns={col: col.upper() for col in df.columns})
print(df.columns)
```

# 🔗 **11. Merge & Concat**

### _`pd.merge(df1, df2, on='col')` → join two DataFrames on a common column | `pd.concat([df1, df2])` → stack DataFrames together_

---

**Q11.1 Create a small lookup DataFrame mapping `origin` codes to full names (`'usa' -> 'United States'`, `'japan' -> 'Japan'`, `'europe' -> 'Europe'`) and merge it with `df` on the `origin` column.**

```python
lookup = pd.DataFrame({"ORIGIN": ["usa", "japan", "europe"], "COUNTRY_NAME": ["United States", "Japan", "Europe"]})
merged_df = pd.merge(df, lookup, on="ORIGIN")
merged_df.head()
```

**Q11.2 Create two smaller DataFrames — the first 5 rows and the last 5 rows of `df` — and combine them back together using `pd.concat()`.**

```python
top5 = df.head(5)
bottom5 = df.tail(5)
concat_df = pd.concat([top5, bottom5], axis=0)
print("Combined shape:", concat_df.shape)
```

**Q11.3 From the merged DataFrame in Q11.1, display only the `name`, `origin`, and the new full-name column for the first 5 rows.**

```python
merged_df[["NAME", "ORIGIN", "COUNTRY_NAME"]].head(5)
```
