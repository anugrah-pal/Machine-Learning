```python
arr = np.arange(10)
print(arr)
np.random.shuffle(arr)
print(arr)
```

```python
#axis = 0 means column wise and axis = 1 means row wise 
arr = np.arange(6).reshape(2, 3)
print(arr)
print(arr.min())
print(arr.min(axis = 0))
print(arr.min(axis = 1))
print(arr.max())
print(arr.max(axis = 0))
print(arr.max(axis = 1))
print(arr.sum())
print(arr.sum(axis = 0))
print(arr.sum(axis = 1))
print(arr.mean())
print(arr.std())
print(np.median(arr))
```

```python
arr = np.arange(24).reshape(4, -1)
print(arr)
```

```python
x = np.arange(10)
arr1 = x[:, np.newaxis]
print(arr1)
arr2 = x[np.newaxis, :]
print(arr2)
```

```python
#FLATTEN - DEEP (FD) & RAVEL - SHALLOW (RS) makes to 1d array from ndarray 
arr = np.arange(12).reshape(4, 3)

arr1 = arr.ravel()
arr1[2] = 99

arr2 = arr.flatten()
arr2[4] = 100

print(arr)
print(arr1)
print(arr2)
```

```python
# np.vstack() and np.hstack - Vertical and Horizontal Stacking 
arr1 = np.array([[1, 3, 4], 
                 [2, 3, 5]
                 ])
arr2 = np.array([[5, 6, 7], 
                 [8, 9, 10], 
                 [10, 11, 14]
                 ])
arr3 = np.arange(12).reshape(2, 6)

arr4 = np.hstack((arr1, arr3))
print(arr4)
arr5 = np.vstack((arr1, arr2))
print(arr5)
```

```python
print(np.concatenate((arr1, arr3), axis = 1))
```

```python
arr = np.full((5, 6), 35)
print(arr)

np.fill_diagonal(arr, 100)
print(arr)
```
