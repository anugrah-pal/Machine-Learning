### Q1.

A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

```python
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])
```

````python
# Q1
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])
bonus = np.array([5, 2, 8])
marks_bonus = marks + bonus
capped_marks = np.minimum(marks_bonus, 100)
num_capped = np.sum(marks_bonus > 100)
print("Capped marks:", capped_marks)
print("Number of marks capped:", num_capped)
````

### Q4.

```python
s1 = np.array([12, 15, 11, 19, 14])
```

### Q13.

A and B are both 2x3 matrices. Compute `A @ B.T` and `A.T @ B`. Print the shape of each result and explain in a comment why both multiplications are legal while `A @ B` is not.

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])
```

````python
# Q13
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])
res1 = A @ B.T
res2 = A.T @ B
print("Shape of A @ B.T:", res1.shape)
print("Shape of A.T @ B:", res2.shape)
# Explanation: In matrix multiplication X @ Y, columns of X must equal rows of Y.
# A (2x3) and B (2x3) cannot be multiplied directly because 3 != 2.
# But A (2x3) @ B.T (3x2) matches (3==3 -> 2x2), and A.T (3x2) @ B (2x3) matches (2==2 -> 3x3).
````

### Q14.

Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by `w = [0.5, 0.2, 0.3]` to obtain one score per row. Finally use `np.where` to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

```python
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
```

````python
# Q14
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
w = np.array([0.5, 0.2, 0.3])
h_col = hours.reshape(-1, 1)
att_col = attendance.reshape(-1, 1)
print("Column vectors shapes:", h_col.shape, att_col.shape)
table2 = np.hstack((h_col, att_col))
print("2-column table shape:", table2.shape)
totals = table2.sum(axis=1, keepdims=True)
table3 = np.hstack((table2, totals))
print("3-column table shape:", table3.shape)
scores = table3 @ w
print("Scores shape:", scores.shape)
labels = np.where(scores > scores.mean(), 1, 0)
print("Labels shape:", labels.shape)
print("Final labels:", labels)
````

### Q16.

Take the array below and create two 1D versions of it — one with `flatten()` and one with `ravel()`. Change the first element of each, then print the original matrix. Explain the difference you observe in a comment.

```python
m = np.array([[4, 8, 15],
              [16, 23, 42]])
```

````python
# Q16
m = np.array([[4, 8, 15],
              [16, 23, 42]])
m_flat = m.flatten()
m_rav = m.ravel()
m_flat[0] = 999
m_rav[0] = 777
print("Original matrix after modifications:", m)
# Explanation: flatten() creates a deep copy in memory so changing m_flat does not affect m.
# ravel() returns a shallow view sharing data with m, so modifying m_rav directly changes m[0, 0].
````

### Q17.

```python
np.random.seed(7)
```

### Q18.

Create two random 3x3 matrices A and B, compute `A @ B` and `B @ A`, and check whether the two results are equal. State your conclusion about matrix multiplication in a comment.

````python
# Q18
np.random.seed(42)
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))
AB = A @ B
BA = B @ A
print("A @ B:", AB)
print("B @ A:", BA)
print("Are they equal?", np.array_equal(AB, BA))
# Conclusion: Matrix multiplication is non-commutative in general (A @ B != B @ A).
````
