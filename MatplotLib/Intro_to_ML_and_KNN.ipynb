```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df = pd.read_csv("../Assignment/Titanic.csv")
```

```python
df.info()
```

```python
# Drop unusable columns, fill age, drop 2 rows
df = df.drop(columns = ['Cabin', 'Name', 'Ticket', 'PassengerId'])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df.dropna(subset='Embarked')
```

```python
# Features and target
x=df.drop(columns=["Survived"])
y=df["Survived"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
# x_train = 80% independent training
# y_train = 80% dependent training
# x_test = 80% independent testing
# y_test = 80% dependent testing
# random_state is same as seed 
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
# ENCODING IN CATEGORICAL DATA 
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
cat_cols = ['Sex', 'Embarked']
```

```python
ct = ColumnTransformer([('cat', OneHotEncoder(sparse_output=False), cat_cols)] ,
                        remainder = 'passthrough')
x_train_final = ct.fit_transform(x_train)
x_test_final = ct.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import  accuracy_score
```

```python
model = KNeighborsClassifier()
model.fit(x_train_final, y_train)
y_pred = model.predict(x_test_final)
print(y_pred)
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

```python
# STANDARDIZATION 
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train_final)
x_test_scale = scaler.transform(x_test_final)
```

```python
scaled_model = KNeighborsClassifier()
scaled_model.fit(x_train_scale, y_train)
y_pred_scale = scaled_model.predict(x_test_scale)
print(y_pred_scale)
accuracy = accuracy_score(y_test, y_pred_scale)
print(accuracy)
```

```python
#HYPER-PARAMETER TUNING 
for k in [3, 5, 9, 11, 15, 19, 21, 25, 29] :
    m = KNeighborsClassifier(n_neighbors=k)
    m.fit(x_train_scale, y_train)
    m_pre = m.predict(x_test_scale)
    print(k, round(accuracy_score(y_test, m_pre), 2))
```
